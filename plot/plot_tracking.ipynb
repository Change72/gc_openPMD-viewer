```python
import matplotlib.pyplot as plt
from datetime import datetime
import numpy as np
import pandas as pd
from matplotlib.ticker import FuncFormatter
from matplotlib.ticker import MaxNLocator, MultipleLocator

import warnings
warnings.filterwarnings('ignore')

plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Times New Roman"]
```

```python
tracking_time_cost = {'particle_num': ['10', '100', '1000'],
                      'time_cost_default': [51.9, 52.5, 52.3],
                      'time_cost_LegoIndex': [0.1, 1.2, 12.4]}

df = pd.DataFrame(tracking_time_cost)
df
```

```python
global_tick_fontsize = 20
global_label_fontsize = 24
global_annotation_fontsize = 20
global_legend_fontsize = 18
```

```python
proj_path = "/data/gc_openPMD-viewer/"
```

```python
fig, ax = plt.subplots(figsize=(6, 4.5))

# Set up the bar chart
bar_width = 0.35  # width of the bars
index = np.arange(len(df['particle_num']))  # bar positions

# Plotting the bars
plt.bar(index, df['time_cost_default'], bar_width, label='Default', color='#bdbdbd')
plt.bar(index + bar_width, df['time_cost_LegoIndex'], bar_width, label='LegoIndex', color='#6baed6')

legend = ax.legend(title='', loc='upper left', labels=['No Index', 'LegoIndex'], fontsize=global_legend_fontsize, ncol=2, frameon=False, bbox_to_anchor=(0.03, 1.18), columnspacing=2.8)
# legend = ax.legend(title='', loc='upper left', labels=['OpenPMD-viewer', 'LegoIndex'], fontsize=global_legend_fontsize, ncol=2, frameon=False, bbox_to_anchor=(-0.1, 1.16), columnspacing=0.8)

for text in legend.get_texts():
    text.set_fontweight('bold')

for i, p in enumerate(ax.patches):
    # if i == 3 or i == 4:
    #     ax.annotate(str(p.get_height()), (p.get_x() + 0.03, p.get_height() + 1), fontsize=global_annotation_fontsize, fontweight='bold')
    # else:
    #     ax.annotate(str(p.get_height()), (p.get_x() + 0.03, p.get_height() + 1), fontsize=global_annotation_fontsize, fontweight='bold')
    ax.annotate(str(p.get_height()), (p.get_x() + 0.03, p.get_height() + 1), fontsize=global_annotation_fontsize, fontweight='bold')

plt.ylim(0, 60)
plt.gca().yaxis.set_major_locator(MultipleLocator(25))
# Add labels and title
plt.xlabel('Tracking Particle Number', fontsize=global_label_fontsize, labelpad=4, fontweight='bold')
plt.ylabel("Avg Time Elapsed (s)", fontsize=global_label_fontsize, fontweight='bold')

plt.gca().tick_params(axis='y', labelsize=global_tick_fontsize)  # Adjust label size as needed
# plt.gca().tick_params(axis='x', labelsize=global_tick_fontsize, pad=2)
plt.gca().tick_params(axis='x', labelsize=global_tick_fontsize, pad=4)
plt.xticks(fontweight='bold')
plt.yticks(fontweight='bold')
# plt.title('Time Cost Comparison: Default vs LegoIndex')
plt.xticks(index + bar_width / 2, df['particle_num'])  # Add particle_num as labels on the x-axis
# plt.legend()

plt.tight_layout()
plt.savefig(proj_path + 'plot/final/tracing_time_cost.pdf', dpi=600, bbox_inches='tight', pad_inches=0.1)
plt.show()
```
