```python
import matplotlib.pyplot as plt
import numpy as np

# Generate example data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x) / 5  # Scaling to keep plot readable

# Create figure and subplots
fig, axes = plt.subplots(nrows=3, ncols=1, sharex=True, figsize=(6, 8))

# Plot data
axes[0].plot(x, y1, label='sin(x)', color='b')
axes[1].plot(x, y2, label='cos(x)', color='r')
axes[2].plot(x, y3, label='tan(x)/5', color='g')

# Add titles and legends
# axes[0].set_title('Sine Function')
# axes[1].set_title('Cosine Function')
# axes[2].set_title('Tangent Function (scaled)')

# Remove space between subplots
plt.subplots_adjust(hspace=0)

for ax in axes:
    ax.legend()
    ax.grid(True)

# Set x-axis label on the last subplot
axes[-1].set_xlabel('X-axis')

# Adjust layout
# plt.tight_layout()
plt.show()
```
